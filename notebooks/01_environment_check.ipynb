```
import torch, torchvision
print("torch:", torch.__version__)
print("torchvision:", torchvision.__version__)
print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("VRAM (MB):", round(torch.cuda.get_device_properties(0).total_memory/1e6, 1))
    print("CUDA version:", torch.version.cuda)
```

```
import sys, json
from pathlib import Path
# Make src importable whether we run from repo root or notebooks/
root = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
sys.path.insert(0, str(root / "src"))

from roadvision.benchmarks.profiler import hardware_report
info = hardware_report(device_used="cpu")
for k, v in info.items():
    print(f"{k:24s}: {v}")
```
