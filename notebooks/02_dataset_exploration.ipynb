# 02 — Dataset Exploration

Index the RDD2022 tree, show class/country statistics, sample images.

**Requires the dataset** (approval-gated). In Colab, mount Drive where the
data lives, or download after `DATASET_APPROVED=true`.

```
import sys, json
from pathlib import Path
root = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
sys.path.insert(0, str(root / "src"))
```

```
from roadvision.config import load_config
config = load_config(root / "configs" / "experiment.yaml")
print("dataset:", config["dataset"]["name"])
print("countries:", config["dataset"]["countries"])
print("approved:", config["dataset"]["approved"])
```

```
from pathlib import Path
raw = root / config["dataset"]["root"]
if not raw.exists() or not any(raw.iterdir()):
    print("Dataset not present. Approval-gated download:")
    print("  1. docs/DATASET_APPROVAL.md approval record")
    print("  2. dataset.approved: true in config")
    print("  3. DATASET_APPROVED=true env var")
    print("  4. python -m roadvision.data.download --countries Czech India Japan")
else:
    print("Data root contents:", sorted(p.name for p in raw.iterdir()))
```

```
# Statistics only run when data exists
if raw.exists() and any(raw.iterdir()):
    from roadvision.data.dataset import make_splits
    splits, manifest = make_splits(raw, config.to_dict())
    print(json.dumps(manifest.to_dict(), indent=2))
```

```
# Optional: preview a few images (needs data)
if raw.exists() and any(raw.iterdir()):
    import subprocess, sys as _s
    _s.call([_s.executable, str(root / "scripts" / "validate_dataset.py"),
             "--config", str(root / "configs" / "experiment.yaml"), "--samples", "6"])
```
